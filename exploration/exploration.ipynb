```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from IPython.display import display

# À adapter aux noms exacts de tes colonnes
DISTANCE = "Distance"
PALETTE = "Palette (1m2) / Encombrement"
POIDS = "poids (tonne)"
CP = "Code Postal"
PRIX = "Prix"

df = df_raw.copy()

# Conversion numérique : accepte les virgules et espaces
for col in [DISTANCE, PALETTE, POIDS, PRIX]:
    df[col] = pd.to_numeric(
        df[col].astype("string")
        .str.replace(r"[\s€]", "", regex=True)
        .str.replace(",", ".", regex=False),
        errors="coerce"
    )

# 57165 → "57" ; conserve les zéros initiaux
df["Departement"] = (
    df[CP].astype("string")
    .str.strip()
    .str.replace(r"\.0$", "", regex=True)
    .str.zfill(5)
    .str[:2]
)

features = [DISTANCE, PALETTE, POIDS]

display(df[features + [PRIX]].describe())
display(
    df[features + [CP, PRIX]]
    .isna().sum().rename("Valeurs manquantes").to_frame()
)

# Exploration : distributions, relations et prix
sns.pairplot(
    df[features + [PRIX]].replace([np.inf, -np.inf], np.nan).dropna(),
    corner=True,
    diag_kind="hist",
    plot_kws={"alpha": 0.4, "s": 20}
)
plt.show()

# Le département est une catégorie : pas une distance numérique.
# Première segmentation sur les caractéristiques physiques.
data = (
    df[features + [PRIX, "Departement"]]
    .replace([np.inf, -np.inf], np.nan)
    .dropna(subset=features)
    .copy()
)

# Standardisation indispensable avant KMeans
X = StandardScaler().fit_transform(data[features])

# Comparaison rapide de plusieurs nombres de groupes
scores = {}
for k in range(2, min(6, len(data) - 1) + 1):
    labels = KMeans(n_clusters=k, random_state=42, n_init=10).fit_predict(X)
    if 1 < len(np.unique(labels)) < len(data):
        scores[k] = silhouette_score(
            X, labels, sample_size=min(3000, len(data)), random_state=42
        )

if not scores:
    raise ValueError("Pas assez de lignes distinctes pour former des groupes.")

display(pd.Series(scores, name="Silhouette").rename_axis("k").to_frame())

k = max(scores, key=scores.get)
data["Cluster"] = KMeans(
    n_clusters=k, random_state=42, n_init=10
).fit_predict(X).astype(str)

# Vue 2D : projection des 3 caractéristiques
pca = PCA(n_components=2)
coords = pca.fit_transform(X)
data["PC1"], data["PC2"] = coords[:, 0], coords[:, 1]

sns.scatterplot(
    data=data, x="PC1", y="PC2", hue="Cluster",
    palette="tab10", alpha=0.6
)
plt.title(
    f"Clusters — PCA "
    f"({pca.explained_variance_ratio_.sum():.0%} de variance conservée)"
)
plt.show()

plt.close("all")

fig = plt.figure(figsize=(9, 6))
ax = fig.add_subplot(111, projection="3d")

ax.scatter(
    data[DISTANCE].to_numpy(dtype=float),
    data[PALETTE].to_numpy(dtype=float),
    data[POIDS].to_numpy(dtype=float),
    c=data["Cluster"].to_numpy(dtype=int),
    cmap="tab10",
    alpha=0.6
)

ax.set_xlabel(DISTANCE)
ax.set_ylabel(PALETTE)
ax.set_zlabel(POIDS)
plt.title("Segmentation logistique")
plt.show()

# Prix : comparaison des groupes et relation avec la distance
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

sns.boxplot(data=data, x="Cluster", y=PRIX, ax=axes[0])
sns.scatterplot(
    data=data, x=DISTANCE, y=PRIX, hue="Cluster",
    palette="tab10", alpha=0.6, ax=axes[1]
)
plt.tight_layout()
plt.show()

# Profil médian des groupes
display(
    data.groupby("Cluster")[features + [PRIX]]
    .median()
    .join(data.groupby("Cluster").size().rename("Nb_lignes"))
)

# Répartition des groupes selon le département
display(
    pd.crosstab(data["Departement"], data["Cluster"], normalize="index")
    .round(2)
)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from IPython.display import display

DISTANCE = "Distance"
PALETTE = "Palette (1m2) / Encombrement"
POIDS = "poids (tonne)"
PRIX = "Prix"

K = 2  # Remets le nombre retenu dans la première étude
FEATURES = [DISTANCE, PALETTE, POIDS]
COLS = FEATURES + [PRIX]

df = df_raw.copy()

for col in COLS:
    df[col] = pd.to_numeric(
        df[col].astype("string")
        .str.replace(r"[\s€]", "", regex=True)
        .str.replace(",", ".", regex=False),
        errors="coerce"
    ).astype(float)

df = (
    df.replace([np.inf, -np.inf], np.nan)
    .dropna(subset=COLS)
    .copy()
)

X = StandardScaler().fit_transform(df[FEATURES])

df["Cluster"] = KMeans(
    n_clusters=K, random_state=42, n_init=10
).fit_predict(X).astype(str)

# Indicatrices des clusters : appartenance = 1, sinon 0
clusters = pd.get_dummies(
    df["Cluster"], prefix="Cluster", dtype=int
)
correlation_data = pd.concat([df[COLS], clusters], axis=1)

# Pearson : relations linéaires
# Spearman : relations monotones, moins sensible aux valeurs extrêmes
fig, axes = plt.subplots(1, 2, figsize=(17, 6))

for ax, method in zip(axes, ["pearson", "spearman"]):
    sns.heatmap(
        correlation_data.corr(method=method),
        annot=True, fmt=".2f",
        cmap="RdBu_r", center=0, vmin=-1, vmax=1,
        ax=ax
    )
    ax.set_title(f"Corrélations — {method.capitalize()}")

plt.tight_layout()
plt.show()

# Relations entre variables, colorées par cluster
sns.pairplot(
    df[COLS + ["Cluster"]],
    hue="Cluster",
    palette="tab10",
    corner=True,
    diag_kind="hist",
    plot_kws={"alpha": 0.4, "s": 20}
)
plt.show()

# Corrélations au sein de chaque cluster
fig, axes = plt.subplots(
    1, K, figsize=(5 * K, 4), squeeze=False
)

for ax, cluster in zip(axes.flat, sorted(df["Cluster"].unique())):
    subset = df.loc[df["Cluster"] == cluster, COLS]

    sns.heatmap(
        subset.corr(method="spearman"),
        annot=True, fmt=".2f",
        cmap="RdBu_r", center=0, vmin=-1, vmax=1,
        ax=ax
    )
    ax.set_title(f"Cluster {cluster} — n={len(subset)}")

plt.tight_layout()
plt.show()

# Corrélation de chaque feature avec le prix : globale vs par cluster
corr_prix = pd.DataFrame({
    "Global": df[COLS].corr(method="spearman")[PRIX].loc[FEATURES]
})

for cluster, subset in df.groupby("Cluster"):
    corr_prix[f"Cluster {cluster}"] = (
        subset[COLS].corr(method="spearman")[PRIX].loc[FEATURES]
    )

display(corr_prix.round(3))
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from IPython.display import display
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# df contient déjà les features nettoyées, Prix et Cluster
data = df.copy()

cluster_encoded = pd.get_dummies(
    data["Cluster"].astype(str),
    prefix="Cluster",
    drop_first=True,
    dtype=int
)

configurations = {
    "1 — Features seules": data[[DISTANCE, PALETTE, POIDS]],
    "2 — Features + cluster": pd.concat(
        [data[[DISTANCE, PALETTE, POIDS]], cluster_encoded], axis=1
    ),
    "3 — Distance + cluster": pd.concat(
        [data[[DISTANCE]], cluster_encoded], axis=1
    )
}

y = data[PRIX]
models = {}
resultats = []

fig, axes = plt.subplots(1, 3, figsize=(16, 5), sharex=True, sharey=True)

for ax, (nom, X) in zip(axes, configurations.items()):
    model = LinearRegression()
    model.fit(X, y)

    prediction = model.predict(X)
    models[nom] = model
    data[f"Prediction — {nom}"] = prediction

    resultats.append({
        "Modèle": nom,
        "R²": r2_score(y, prediction),
        "MAE": mean_absolute_error(y, prediction),
        "RMSE": np.sqrt(mean_squared_error(y, prediction))
    })

    print(f"\n{nom}")
    display(pd.Series(
        [model.intercept_, *model.coef_],
        index=["Constante", *X.columns],
        name="Coefficient"
    ).to_frame().round(3))

    ax.scatter(y, prediction, alpha=0.5, s=20)
    bornes = [min(y.min(), prediction.min()), max(y.max(), prediction.max())]
    ax.plot(bornes, bornes, "--", color="black")
    ax.set_title(nom)
    ax.set_xlabel("Prix réel")
    ax.set_ylabel("Prix prédit")

plt.tight_layout()
plt.show()

display(pd.DataFrame(resultats).set_index("Modèle").round(3))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from IPython.display import display

# ===== DATASET — tarifs 24T HT du tableau =====
df_zones = pd.DataFrame({
    "Departement": ["12", "13", "15", "18", "19", "79", "86"],
    "Distance_km": [600, 823, 520, 210, 431, 366, 300],
    "Prix_HT": [1600, 2100, 1600, 725, 1400, 850, 725]
})

df_zones["Prix_par_km"] = df_zones["Prix_HT"] / df_zones["Distance_km"]

X = df_zones[["Distance_km"]]
y = df_zones["Prix_HT"]

# ===== RÉGRESSIONS PAR MOINDRES CARRÉS =====
modeles = {
    "Tarif kilométrique seul": LinearRegression(fit_intercept=False),
    "Forfait + tarif kilométrique": LinearRegression(fit_intercept=True)
}

resultats = []

for nom, modele in modeles.items():
    modele.fit(X, y)
    predictions = modele.predict(X)

    resultats.append({
        "Modèle": nom,
        "Alpha (€)": float(modele.intercept_),
        "Beta (€/km)": modele.coef_[0],
        "R²": r2_score(y, predictions),
        "MAE (€)": mean_absolute_error(y, predictions),
        "RMSE (€)": np.sqrt(mean_squared_error(y, predictions))
    })

comparaison = pd.DataFrame(resultats).set_index("Modèle")

# Meilleur ajustement sur ces 7 observations
meilleur_nom = comparaison["RMSE (€)"].idxmin()
meilleur_modele = modeles[meilleur_nom]

df_zones["Prix_predit"] = meilleur_modele.predict(X)
df_zones["Ecart_reel_moins_predit"] = (
    df_zones["Prix_HT"] - df_zones["Prix_predit"]
)

display(df_zones.round(2))
display(comparaison.round(4))

print(f"Meilleur ajustement : {meilleur_nom}")
print(
    f"Prix HT = {float(meilleur_modele.intercept_):.2f}"
    f" + {meilleur_modele.coef_[0]:.4f} × distance"
)

# ===== SCATTER + DROITES =====
distances = pd.DataFrame({
    "Distance_km": np.linspace(0, df_zones["Distance_km"].max() * 1.05, 200)
})

fig, ax = plt.subplots(figsize=(10, 6))

ax.scatter(
    df_zones["Distance_km"],
    df_zones["Prix_HT"],
    color="black", s=65, zorder=3,
    label="Tarifs observés"
)

for _, ligne in df_zones.iterrows():
    ax.annotate(
        f"Dept {ligne['Departement']}",
        (ligne["Distance_km"], ligne["Prix_HT"]),
        xytext=(6, 8), textcoords="offset points"
    )

for nom, modele in modeles.items():
    ax.plot(
        distances["Distance_km"],
        modele.predict(distances),
        linestyle="-" if nom == meilleur_nom else "--",
        label=(
            f"{nom} : "
            f"{float(modele.intercept_):.1f} € "
            f"+ {modele.coef_[0]:.3f} €/km × distance"
        )
    )

ax.set_xlabel("Distance (km)")
ax.set_ylabel("Prix 24T HT (€)")
ax.set_title("Tarifs par département selon la distance")
ax.legend()
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()

# Une seule observation par département :
# impossible d'isoler un effet propre à chaque zone avec ces seules données.
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error
from IPython.display import display

# ===== DATASET =====
df_zones = pd.DataFrame({
    "Departement": ["12", "13", "15", "18", "19", "79", "86"],
    "Distance_km": [600, 823, 520, 210, 431, 366, 300],
    "Prix_HT": [1600, 2100, 1600, 725, 1400, 850, 725]
})

# Groupe 1 : 18, 19, 15 ; groupe 2 : le reste
df_zones["Groupe_2"] = (
    ~df_zones["Departement"].isin(["18", "19", "15"])
).astype(int)

# ===== AJUSTEMENT CONJOINT : PENTE COMMUNE =====
X = df_zones[["Distance_km", "Groupe_2"]]
y = df_zones["Prix_HT"]

modele = LinearRegression().fit(X, y)

A = modele.coef_[0]
b1 = modele.intercept_
b2 = b1 + modele.coef_[1]

df_zones["Prix_predit"] = modele.predict(X)
df_zones["Residu"] = y - df_zones["Prix_predit"]

display(pd.DataFrame({
    "Groupe": ["1 : 18, 19, 15", "2 : 12, 13, 79, 86"],
    "A commun (€/km)": [A, A],
    "Forfait b (€)": [b1, b2]
}).round(3))

print(f"Groupe 1 : Prix = {A:.4f} × distance + ({b1:.2f})")
print(f"Groupe 2 : Prix = {A:.4f} × distance + ({b2:.2f})")
print(f"Décalage groupe 1 − groupe 2 : {b1 - b2:.2f} €")
print(f"R² global : {r2_score(y, df_zones['Prix_predit']):.4f}")
print(f"MAE globale : {mean_absolute_error(y, df_zones['Prix_predit']):.2f} €")

display(df_zones.round(2))

# ===== SCATTER + DROITES PARALLÈLES =====
fig, ax = plt.subplots(figsize=(10, 6))

for groupe, b, couleur, nom in [
    (0, b1, "tab:orange", "Groupe 1 : 18, 19, 15"),
    (1, b2, "tab:blue", "Groupe 2 : le reste")
]:
    subset = df_zones.loc[df_zones["Groupe_2"] == groupe]

    ax.scatter(
        subset["Distance_km"], subset["Prix_HT"],
        color=couleur, s=70, label=nom, zorder=3
    )

    distances = np.linspace(
        subset["Distance_km"].min(),
        subset["Distance_km"].max(),
        100
    )
    ax.plot(
        distances, A * distances + b,
        color=couleur,
        label=f"Prix = {A:.3f} × distance + ({b:.1f})"
    )

    for _, ligne in subset.iterrows():
        ax.annotate(
            f"Dept {ligne['Departement']}",
            (ligne["Distance_km"], ligne["Prix_HT"]),
            xytext=(6, 8), textcoords="offset points"
        )

ax.set_xlabel("Distance (km)")
ax.set_ylabel("Prix 24T HT (€)")
ax.set_title("Deux forfaits avec une pente kilométrique commune")
ax.legend()
ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
from IPython.display import display


class ModeleLivraison:
    def __init__(self):
        self.slopes_zones = {
            "15": 3.1736,
            "18": 3.1736,
            "19": 3.1736,
            "12": 2.5505,
            "13": 2.5505,
            "79": 2.5505,
            "86": 2.5505,
        }
        self.modele_base = LinearRegression(fit_intercept=False)

    @staticmethod
    def departements(codes_postaux):
        codes = (
            codes_postaux.astype("string")
            .str.strip()
            .str.replace(r"\s+", "", regex=True)
            .str.replace(r"[.,]0+$", "", regex=True)
            .str.zfill(5)
        )

        # Codes invalides → NA ; 57155 → "57" ; 1000 → "01"
        codes = codes.where(codes.str.fullmatch(r"\d{5}", na=False))
        return codes.str[:2]

    def fit(self, data, distance_col, prix_col, cp_col):
        self.distance_col = distance_col
        self.cp_col = cp_col

        departements = self.departements(data[cp_col])
        hors_zone = (
            departements.notna()
            & ~departements.isin(self.slopes_zones)
        )

        if not hors_zone.any():
            raise ValueError("Aucune livraison hors zone pour entraîner la base.")

        if not data.loc[hors_zone, distance_col].gt(0).any():
            raise ValueError("Aucune distance positive hors zone.")

        # Prix = slope_base × distance, origine forcée à 0
        self.modele_base.fit(
            data.loc[hors_zone, [distance_col]],
            data.loc[hors_zone, prix_col]
        )

        self.slope_base = float(self.modele_base.coef_[0])
        self.nb_train = int(hors_zone.sum())
        return self

    def predict(self, data):
        resultat = data.copy()
        resultat["Departement"] = self.departements(resultat[self.cp_col])

        if resultat["Departement"].isna().any():
            raise ValueError("Code postal invalide dans les données à prédire.")

        resultat["Slope"] = (
            resultat["Departement"]
            .map(self.slopes_zones)
            .fillna(self.slope_base)
        )

        resultat["Tarif"] = np.where(
            resultat["Departement"].isin(self.slopes_zones),
            "Zone difficile",
            "Base"
        )

        resultat["Prix_predit"] = (
            resultat[self.distance_col] * resultat["Slope"]
        )

        return resultat


# ===== PRÉPARATION =====
DISTANCE = "Distance"
PRIX = "Prix"
CP = "Code Postal"

data = df_raw.copy()
nb_initial = len(data)

for col in [DISTANCE, PRIX]:
    data[col] = pd.to_numeric(
        data[col].astype("string")
        .str.replace(r"[\s€]", "", regex=True)
        .str.replace(",", ".", regex=False),
        errors="coerce"
    ).astype(float)

# Suppression des codes postaux manquants ou invalides
departements = ModeleLivraison.departements(data[CP])
data = data.loc[departements.notna()].copy()

# Suppression des distances/prix manquants ou infinis
data = (
    data.replace({DISTANCE: {np.inf: np.nan, -np.inf: np.nan},
                  PRIX: {np.inf: np.nan, -np.inf: np.nan}})
    .dropna(subset=[DISTANCE, PRIX])
    .loc[lambda x: x[DISTANCE].ge(0)]
    .copy()
)

print(f"Lignes supprimées au nettoyage : {nb_initial - len(data)}")

# ===== ENTRAÎNEMENT UNIQUEMENT HORS ZONES DIFFICILES =====
modele = ModeleLivraison().fit(data, DISTANCE, PRIX, CP)

print(f"Livraisons utilisées pour la base : {modele.nb_train}")
print(f"Slope de base : {modele.slope_base:.4f} €/km")

# ===== PRÉDICTIONS SUR TOUT LE DATASET NETTOYÉ =====
resultats = modele.predict(data)
resultats["Residu"] = resultats[PRIX] - resultats["Prix_predit"]

display(resultats[
    [CP, "Departement", DISTANCE, PRIX, "Tarif",
     "Slope", "Prix_predit", "Residu"]
].round(2))

# ===== ÉVALUATION =====
scores = []

for nom, subset in [
    ("Global", resultats),
    *list(resultats.groupby("Tarif"))
]:
    scores.append({
        "Périmètre": nom,
        "Nb livraisons": len(subset),
        "MAE": mean_absolute_error(subset[PRIX], subset["Prix_predit"]),
        "RMSE": np.sqrt(
            mean_squared_error(subset[PRIX], subset["Prix_predit"])
        )
    })

display(pd.DataFrame(scores).set_index("Périmètre").round(2))

# ===== TEST MANUEL : MODIFIE ICI =====
nouvelle_livraison = pd.DataFrame({
    DISTANCE: [150],
    CP: ["57155"]
})

display(modele.predict(nouvelle_livraison).round(2))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.cluster.hierarchy import linkage, dendrogram, cut_tree
from sklearn.metrics import silhouette_score
from IPython.display import display

visu = df_calibration.copy()
visu[POIDS] = pd.to_numeric(visu[POIDS], errors="coerce")
visu = (
    visu.replace({POIDS: {np.inf: np.nan, -np.inf: np.nan}})
    .dropna(subset=[POIDS])
    .copy()
)

X = visu[[POIDS]].to_numpy(dtype=float)
n = len(X)

if n < 3 or visu[POIDS].nunique() < 2:
    raise ValueError("Pas assez de poids distincts pour comparer les groupes.")

# ===== CAH : POIDS UNIQUEMENT, MÉTHODE DE WARD =====
Z = linkage(X, method="ward")

# ===== COMPARAISON DE 2 À 8 GROUPES =====
scores = []
labels_par_k = {}

for k in range(2, min(8, n - 1, visu[POIDS].nunique()) + 1):
    labels = cut_tree(Z, n_clusters=k).ravel()
    labels_par_k[k] = labels

    scores.append({
        "Nb_clusters": k,
        "Silhouette": silhouette_score(
            X,
            labels,
            sample_size=min(3000, n),
            random_state=42
        )
    })

comparaison = pd.DataFrame(scores)
BEST_K = int(
    comparaison.loc[comparaison["Silhouette"].idxmax(), "Nb_clusters"]
)

display(comparaison.round(4))
print(f"Meilleur nombre de groupes selon la silhouette : {BEST_K}")

# ===== DENDROGRAMME + SCORE =====
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

dendrogram(
    Z,
    truncate_mode="lastp",
    p=min(30, n),
    show_leaf_counts=True,
    ax=axes[0]
)
axes[0].set_title("CAH sur le poids — dendrogramme tronqué")
axes[0].set_xlabel("Observations ou groupes fusionnés")
axes[0].set_ylabel("Distance de fusion — Ward")

axes[1].plot(
    comparaison["Nb_clusters"],
    comparaison["Silhouette"],
    marker="o"
)
axes[1].axvline(BEST_K, color="red", linestyle="--")
axes[1].set_xlabel("Nombre de clusters")
axes[1].set_ylabel("Score de silhouette")
axes[1].set_title("Plus le score est élevé, mieux les groupes sont séparés")
axes[1].grid(alpha=0.25)

plt.tight_layout()
plt.show()

# ===== PRIX / POIDS POUR CHAQUE DÉCOUPAGE =====
fig, axes = plt.subplots(2, 4, figsize=(20, 9), sharex=True, sharey=True)

for ax, (k, labels) in zip(axes.flat, labels_par_k.items()):
    sns.scatterplot(
        x=visu[POIDS].to_numpy(),
        y=visu[PRIX].to_numpy(),
        hue=labels.astype(str),
        palette="tab10",
        s=35,
        alpha=0.7,
        ax=ax
    )
    ax.set_title(f"{k} clusters")
    ax.set_xlabel(POIDS)
    ax.set_ylabel(PRIX)
    ax.legend(title="Cluster", fontsize=7)

for ax in list(axes.flat)[len(labels_par_k):]:
    ax.set_visible(False)

plt.tight_layout()
plt.show()

# ===== DÉCOUPAGE RETENU =====
# Numérotation du groupe le plus léger au plus lourd
visu["Cluster_poids"] = labels_par_k[BEST_K]

ordre = (
    visu.groupby("Cluster_poids")[POIDS]
    .mean()
    .sort_values()
    .index
)
mapping = {ancien: nouveau for nouveau, ancien in enumerate(ordre)}
visu["Cluster_poids"] = visu["Cluster_poids"].map(mapping)

display(
    visu.groupby("Cluster_poids").agg(
        Effectif=(POIDS, "size"),
        Poids_min=(POIDS, "min"),
        Poids_median=(POIDS, "median"),
        Poids_max=(POIDS, "max"),
        Prix_median=(PRIX, "median")
    ).round(2)
)

# Les résultats sont disponibles dans :
# visu["Cluster_poids"]
```

```python
import math
import pandas as pd
from IPython.display import display

STRATES_POIDS = {
    0: {"max": 5.5, "coef_poids": 48.7581},
    1: {"max": 12,  "coef_poids": 34.3832},
    2: {"max": 21,  "coef_poids": 35.8118},
    3: {"max": 28,  "coef_poids": 35.1261},
}

COEFS_DISTANCE = {
    "standard": 0.966,
    "15": 3.1736,
    "18": 3.1736,
    "19": 3.1736,
    "12": 2.5505,
    "13": 2.5505,
    "79": 2.5505,
    "86": 2.5505,
}

CAPACITE_CAMION = 28.0


def predire_livraison(poids_total, distance, code_postal):
    poids_total = float(poids_total)
    distance = float(distance)

    if not math.isfinite(poids_total) or poids_total <= 0:
        raise ValueError("Le poids total doit être strictement positif.")
    if not math.isfinite(distance) or distance < 0:
        raise ValueError("La distance doit être positive ou nulle.")

    # Normalisation du code postal : 1000 → 01000
    code = str(code_postal).strip()
    if code.endswith(".0"):
        code = code[:-2]
    code = code.zfill(5)

    if len(code) != 5 or not code.isascii() or not code.isdigit():
        raise ValueError("Code postal invalide.")

    departement = code[:2]
    coef_distance = COEFS_DISTANCE.get(
        departement, COEFS_DISTANCE["standard"]
    )

    # Camions pleins, puis éventuel dernier camion partiel
    nb_pleins = int(poids_total // CAPACITE_CAMION)
    reste = poids_total % CAPACITE_CAMION

    poids_camions = [CAPACITE_CAMION] * nb_pleins
    if reste > 0:
        poids_camions.append(reste)

    lignes = []

    for numero, poids in enumerate(poids_camions, start=1):
        strate = next(
            s for s, regle in STRATES_POIDS.items()
            if poids <= regle["max"]
        )
        coef_poids = STRATES_POIDS[strate]["coef_poids"]

        part_poids = poids * coef_poids
        part_distance = distance * coef_distance

        lignes.append({
            "Camion": numero,
            "ZIP": code,
            "Departement": departement,
            "Poids_tonnes": poids,
            "Strate": strate,
            "Distance_km": distance,
            "Coef_poids": coef_poids,
            "Coef_distance": coef_distance,
            "Part_poids": part_poids,
            "Part_distance": part_distance,
            "Prix_camion": part_poids + part_distance
        })

    detail = pd.DataFrame(lignes)
    prix_total = float(detail["Prix_camion"].sum())

    return detail, prix_total


# ===== TEST MANUEL =====
detail, prix_total = predire_livraison(
    poids_total=2.117	,       # 28 + 28 + 28 + 14 tonnes
    distance=700,
    code_postal="31000"
)

display(detail.round(2))
print(f"Nombre de camions : {len(detail)}")
print(f"Prix total de la livraison : {prix_total:.2f} €")
```

```python
from lightgbm import LGBMRegressor

# Noms simples pour éviter les modifications de noms par LightGBM
def features_correction(data):
    return pd.DataFrame({
        "distance": data[DISTANCE].to_numpy(dtype=float),
        "poids": data[POIDS].to_numpy(dtype=float),
        "strate": data["Strate"].to_numpy(dtype=int),
        "coef_distance": data["Coef_distance"].to_numpy(dtype=float),
        "prix_base": data["Prix_base"].to_numpy(dtype=float)
    }, index=data.index)

X_correction = features_correction(df_residus)
y_correction = df_residus["Residu_cible"]

PARAMS_CORRECTION = {
    "objective": "regression",
    "n_estimators": 1000,
    "learning_rate": 0.03,
    "num_leaves": 8,
    "max_depth": 10,
    "min_child_samples": 5,
    "reg_lambda": 1.0,
    "random_state": 42,
    "verbosity": -1
}

modele_correction = LGBMRegressor(**PARAMS_CORRECTION)
modele_correction.fit(X_correction, y_correction)

df_residus["Correction_LGBM"] = modele_correction.predict(X_correction)

df_residus["Prix_hybride"] = (
    df_residus["Prix_base"] + df_residus["Correction_LGBM"]
)
```

```python
from sklearn.model_selection import KFold, cross_val_predict
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

cv = KFold(
    n_splits=min(5, len(df_residus)),
    shuffle=True,
    random_state=42
)

correction_cv = cross_val_predict(
    LGBMRegressor(**PARAMS_CORRECTION),
    X_correction,
    y_correction,
    cv=cv
)

df_residus["Prix_hybride_CV"] = (
    df_residus["Prix_base"] + correction_cv
)

comparaison = []

for nom, colonne in {
    "Coefficients seuls": "Prix_base",
    "Coefficients + LGBM — fit": "Prix_hybride",
    "Coefficients + LGBM — CV": "Prix_hybride_CV"
}.items():
    reel = df_residus[PRIX]
    predit = df_residus[colonne]
    erreur_absolue = (predit - reel).abs()

    comparaison.append({
        "Modèle": nom,
        "MAE (€)": mean_absolute_error(reel, predit),
        "RMSE (€)": np.sqrt(mean_squared_error(reel, predit)),
        "MAPE (%)": (
            100 * erreur_absolue / reel.abs().replace(0, np.nan)
        ).mean(),
        "R²": r2_score(reel, predit)
    })

display(
    pd.DataFrame(comparaison)
    .set_index("Modèle")
    .round(3)
)

df_residus["Erreur_hybride_CV_pct"] = (
    100 * (df_residus["Prix_hybride_CV"] - df_residus[PRIX]).abs()
    / df_residus[PRIX].abs().replace(0, np.nan)
)

display(
    df_residus[
        [CP, DISTANCE, POIDS, PRIX,
         "Prix_base", "Prix_hybride", "Prix_hybride_CV",
         "Erreur_base_pct", "Erreur_hybride_CV_pct"]
    ].round(2)
)
```

```python
import re
import numpy as np
import pandas as pd
from IPython.display import display

DISTANCE = "Distance"
POIDS = "poids (tonne)"
PALETTE = "Palette (1m2) / Encombrement"
CP = "Code Postal"
PRIX = "Prix"

resultats_hybrides = df_raw.copy()

# ===== NETTOYAGE NUMÉRIQUE =====
for col in [DISTANCE, POIDS, PALETTE, PRIX]:
    resultats_hybrides[col] = pd.to_numeric(
        resultats_hybrides[col].astype("string")
        .str.replace(r"[\s€]", "", regex=True)
        .str.replace(",", ".", regex=False),
        errors="coerce"
    ).astype(float)


def normaliser_localisation(valeur, departement_seul=False):
    if pd.isna(valeur):
        return None

    texte = re.sub(r"\s+", "", str(valeur).strip())
    texte = re.sub(r"[.,]0+$", "", texte)

    if not re.fullmatch(r"[0-9]+", texte):
        return None

    if departement_seul:
        return texte.zfill(2) if len(texte) <= 2 else None

    if len(texte) == 2:
        # Département déjà renseigné, exemple : "13"
        return texte

    if len(texte) in (4, 5):
        # Code postal complet : 1000 → 01000 → département 01
        return texte.zfill(5)[:2]

    return None


# ===== INFÉRENCE SUR TOUTES LES LIGNES, TOUTES ZONES =====
predictions = []

for _, ligne in resultats_hybrides.iterrows():
    departement = normaliser_localisation(ligne[CP])

    # Si le ZIP est inexploitable, essaie une colonne département existante
    if departement is None:
        departement = normaliser_localisation(
            ligne.get("Departement", pd.NA),
            departement_seul=True
        )

    try:
        if departement is None:
            raise ValueError(
                f"Localisation inexploitable : {ligne[CP]!r}"
            )

        # La fonction utilise uniquement les 2 premiers chiffres du ZIP.
        # Ce code technique transmet donc le département déjà identifié.
        code_pour_modele = departement + "000"

        resume, _ = predire_prix_hybride(
            distance=ligne[DISTANCE],
            poids=ligne[POIDS],
            palettes=ligne[PALETTE],
            code_postal=code_pour_modele
        )

        predictions.append({
            "Departement_modele": departement,
            "Coef_distance": COEFS_DISTANCE.get(
                departement, COEFS_DISTANCE["standard"]
            ),
            "Nb_camions": resume["Nb_camions"],
            "Prix_base": resume["Prix_base"],
            "Correction_LGBM": resume["Correction_LGBM"],
            "Prix_estime": resume["Prix_hybride"],
            "Statut": "OK"
        })

    except (ValueError, TypeError, OverflowError) as erreur:
        predictions.append({
            "Departement_modele": departement,
            "Coef_distance": np.nan,
            "Nb_camions": pd.NA,
            "Prix_base": np.nan,
            "Correction_LGBM": np.nan,
            "Prix_estime": np.nan,
            "Statut": str(erreur)
        })

predictions = pd.DataFrame(predictions)

for col in predictions.columns:
    resultats_hybrides[col] = predictions[col].to_numpy()

resultats_hybrides["Nb_camions"] = (
    resultats_hybrides["Nb_camions"].astype("Int64")
)

# ===== ERREURS =====
valides = (
    np.isfinite(resultats_hybrides[PRIX])
    & np.isfinite(resultats_hybrides["Prix_estime"])
)

resultats_hybrides["Ecart"] = (
    resultats_hybrides["Prix_estime"] - resultats_hybrides[PRIX]
).where(valides)

resultats_hybrides["Erreur_absolue"] = (
    resultats_hybrides["Ecart"].abs()
)

resultats_hybrides["Erreur_absolue_pct"] = (
    100 * resultats_hybrides["Erreur_absolue"]
    / resultats_hybrides[PRIX].abs().replace(0, np.nan)
)

evaluation = resultats_hybrides.loc[valides]

display(pd.Series({
    "Lignes totales": len(resultats_hybrides),
    "Lignes prédites": resultats_hybrides["Statut"].eq("OK").sum(),
    "Lignes évaluées pour MAE": len(evaluation),
    "Lignes évaluées pour MAPE":
        evaluation["Erreur_absolue_pct"].notna().sum(),
    "MAE (€)": evaluation["Erreur_absolue"].mean(),
    "MAPE (%)": evaluation["Erreur_absolue_pct"].mean()
}, name="Modèle hybride").to_frame().round(2))

display(
    resultats_hybrides[
        [CP, "Departement_modele", "Coef_distance",
         DISTANCE, POIDS, PALETTE, "Nb_camions",
         PRIX, "Prix_base", "Correction_LGBM", "Prix_estime",
         "Ecart", "Erreur_absolue_pct", "Statut"]
    ].round(2)
)
```
